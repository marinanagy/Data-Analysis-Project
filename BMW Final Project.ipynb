```python
top_countries=df.groupby("country")["Revenue"].sum().sort_values(ascending=False).head(6)

plt.figure(figsize=(10,6))

top_countries.plot(kind='bar')

plt.title("Top 6 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Total Revenue")

plt.show()
```

```python
top_revenue_after_marketing=df.groupby("country")["Revenue After Marketing"].sum().sort_values(ascending=False).head(6)

plt.figure(figsize=(10,6))

top_countries.plot(kind='bar')

plt.title("Revenue After Marketing")
plt.xlabel("Country")
plt.ylabel("Total Revenue After Marketing")

plt.show()
```

```python
revenue_per_year = df.groupby('year')["Revenue"].sum().reset_index()

plt.figure(figsize=(10,6))

sns.lineplot(
    data=revenue_per_year,
    x='year',
    y='Revenue',
    marker='o'
)

plt.title("Revenue Per Year")
plt.xlabel("Year")
plt.ylabel("Total Revenue")

plt.show()
```

```python
model_revenue=df.groupby('model')['Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))

sns.barplot(
    x=model_revenue.index,
    y=model_revenue.values
)

plt.title("Revenue by BMW Model")
plt.xlabel("Model")
plt.ylabel("Total Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
engine_type_revenue=df.groupby('engine_type')['Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,8))

plt.pie(
    engine_type_revenue,
    labels=engine_type_revenue.index,
    autopct='%1.1f%%'
)

plt.title("Revenue Per Engine")

plt.show()
```

```python
segment_revenue=df.groupby('segment')['Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,8))

plt.pie(
    segment_revenue,
    labels=segment_revenue.index,
    autopct='%1.1f%%'
)

plt.title("Revenue Per segment")

plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='Purples')

plt.title("Correlation Matrix")
plt.show()
```
